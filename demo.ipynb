# Demostración del flujo de delegación

La consulta exige investigación documental, cálculo de prioridad, validación y síntesis.

```python
from multi_agent_orchestrator.runner import run_orchestrator

query = (
    "Investiga el impacto del aumento de errores en checkout, calcula su prioridad "
    "y propone acciones inmediatas basadas en evidencia."
)
state, trace = await run_orchestrator(query)
```

```python
for event in trace:
    if event["node"] == "supervisor":
        print(f"supervisor -> {event['decision']}")
    elif event["node"] == "researcher":
        print(f"researcher -> {len(event['sources'])} fuentes recuperadas")
    elif event["node"] == "analyst":
        result = event["analysis"]
        print(f"analyst -> {result['priority']}, riesgo {result['risk_score']}")
    elif event["node"] == "validator":
        status = "aprobado" if event["task_completed"] else "requiere ajustes"
        print(f"validator -> {status}")
    else:
        print("synthesizer -> respuesta final")
```
